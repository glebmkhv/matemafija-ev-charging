# MATEMAFIJA history exploration

Portable, output-free starting point for exploring the local SQLite history. The working database is intentionally not committed; create it with `python src/collect.py --once` from the repository root. Observed, reconstructed, estimated, assumed and simulated values must remain explicitly separated.

```python
from pathlib import Path
import sqlite3

repo_root = Path.cwd()
if not (repo_root / 'src').exists():
    repo_root = repo_root.parent
db_path = repo_root / 'data' / 'ev_charging.db'
db_path
```

```python
if not db_path.exists():
    raise FileNotFoundError(
        'Local database not found. Run: python src/collect.py --once'
    )

with sqlite3.connect(f'file:{db_path}?mode=ro', uri=True) as connection:
    tables = connection.execute(
        "SELECT name FROM sqlite_master WHERE type = 'table' ORDER BY name"
    ).fetchall()

[name for (name,) in tables]
```
